```
import numpy as np
import matplotlib.pyplot as plt

# Sample Data (two classes)
np.random.seed(0)
X1 = np.random.randn(50, 2) + np.array([2, 2])  # Class 1
X2 = np.random.randn(50, 2) + np.array([-2, -2]) # Class 2
X = np.concatenate([X1, X2])
y = np.concatenate([np.zeros(50), np.ones(50)])  # 0 for Class 1, 1 for Class 2

# Decision Function (Linear Example)
def decision_function(x):
    w = np.array([1, 1])  # Weights
    b = -1             # Bias
    return np.dot(x, w) + b

# Plotting
plt.figure(figsize=(8, 6))
plt.scatter(X[y == 0, 0], X[y == 0, 1], label='Class 1', marker='o')
plt.scatter(X[y == 1, 0], X[y == 1, 1], label='Class 2', marker='x')

# Plot the Decision Boundary
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),
                     np.arange(y_min, y_max, 0.01))
Z = decision_function(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
plt.contourf(xx, yy, Z, levels=0, colors=('b', 'r'), alpha=0.2) # Filled contour

# Highlight the decision boundary line
plt.contour(xx, yy, Z, levels=[0], linewidths=2, colors='black')

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Decision Boundary')
plt.legend()
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.show()


# Example of classifying a new point:
new_point = np.array([0, 1])
decision_value = decision_function(new_point)

if decision_value > 0:
    print(f"The point {new_point} is classified as Class 1")
else:
    print(f"The point {new_point} is classified as Class 2")

#Non-linear Example (circle)

def decision_function_nonlinear(x):
  center = np.array([0,0])
  radius = 2
  # If x is 1-dimensional, calculate the norm directly
  if x.ndim == 1:
    distance_from_center = np.linalg.norm(x - center)
  else:
    distance_from_center = np.linalg.norm(x - center, axis=1)
  return radius - distance_from_center #Positive if inside, negative if outside

plt.figure(figsize=(8, 6))
plt.scatter(X[y == 0, 0], X[y == 0, 1], label='Class 1', marker='o')
plt.scatter(X[y == 1, 0], X[y == 1, 1], label='Class 2', marker='x')

x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),
                     np.arange(y_min, y_max, 0.01))
Z = decision_function_nonlinear(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
plt.contourf(xx, yy, Z, levels=0, colors=('b', 'r'), alpha=0.2) # Filled contour

# Highlight the decision boundary line
plt.contour(xx, yy, Z, levels=[0], linewidths=2, colors='black')

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Non-linear Decision Boundary')
plt.legend()
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.show()

new_point = np.array([1, 1])
decision_value = decision_function_nonlinear(new_point)

if decision_value > 0:
    print(f"The point {new_point} is classified as Class 1 (inside)")
else:
    print(f"The point {new_point} is classified as Class 2 (outside)")

print(f"\nSum of posterior probabilities: {total_posterior_prob:.3f}") #Should be very close to 1.0
```
