```
import numpy as np
import statsmodels.api as sm
from sklearn.metrics import log_loss

# Dataset
hours_studied = np.array([2, 3, 5, 7, 8])
passed_exam = np.array([0, 0, 1, 1, 1])

# Add a constant term (intercept) to the input data
X = sm.add_constant(hours_studied)

# Create the logistic regression model with a higher maximum number of iterations
# and using method=‘bfgs’ for better handling near-singularities
model = sm.Logit(passed_exam, X)
result = model.fit(method='bfgs', maxiter=1000)

# Print the coefficients
print(result.params)

# Make predictions
pred_prob = result.predict(X)

# Calculate log-loss
loss = log_loss(passed_exam, pred_prob)

print(f'Log-Loss: {loss}')
```
