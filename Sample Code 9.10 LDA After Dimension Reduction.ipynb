```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# Generate Synthetic Dataset with 3 Classes
# Reduced n_informative to 3 to be less than n_features
X, y = make_classification(
    n_samples=300, n_features=5, n_informative=3, n_classes=3, random_state=42
)

# Apply Linear Discriminant Analysis (LDA)
lda = LinearDiscriminantAnalysis(n_components=2)  # Reduce to 2 dimensions
X_lda = lda.fit_transform(X, y)

# Plot the reduced dimensions
plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_lda[:, 0], X_lda[:, 1], c=y, cmap='viridis', edgecolor='k', alpha=0.7)
plt.title("LDA Dimensionality Reduction to 2 Components", fontsize=16)
plt.xlabel("LD1 (Linear Discriminant 1)", fontsize=14)
plt.ylabel("LD2 (Linear Discriminant 2)", fontsize=14)
plt.colorbar(scatter, label="Class Labels")
plt.grid(alpha=0.4)
plt.show()

# Print Explained Variance Ratio
print("Explained Variance Ratio:", lda.explained_variance_ratio_)
```
