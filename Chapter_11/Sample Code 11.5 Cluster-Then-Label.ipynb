```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

# Step 1: Generate Synthetic Data
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=1.2, random_state=42)
scaler = StandardScaler()
X = scaler.fit_transform(X)  # Normalize features for better clustering

# Step 2: Perform Unsupervised Clustering (K-Means)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X)

# Step 3: Manually Assign Labels to Each Cluster
# Assume we have labeled a few points per cluster (simulating human annotation)
cluster_to_class = {
    0: 1,  # Assign cluster 0 to class 1
    1: 0,  # Assign cluster 1 to class 0
    2: 2   # Assign cluster 2 to class 2
}

# Step 4: Assign Labels to the Entire Dataset Based on Cluster Membership
y_pred = np.array([cluster_to_class[cluster] for cluster in cluster_labels])

# Step 5: Plot Original Data vs. Cluster-Then-Label Results
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Original Data
axes[0].scatter(X[:, 0], X[:, 1], c=y_true, cmap='coolwarm', edgecolors='k', alpha=0.7)
axes[0].set_title("Original Data (True Labels)")

# Cluster-Then-Label Results
axes[1].scatter(X[:, 0], X[:, 1], c=y_pred, cmap='coolwarm', edgecolors='k', alpha=0.7)
axes[1].set_title("Cluster-Then-Label Results")

plt.show()
```
