```

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.semi_supervised import LabelPropagation

# Step 1: Generate the Half-Moons Dataset
X, y = make_moons(n_samples=300, noise=0.2, random_state=42)  # 300 samples

# Step 2: Limit Labeled Data (Only 10 Labeled Points)
rng = np.random.RandomState(42)
unlabeled_indices = rng.choice(len(y), size=290, replace=False)  # Select 290 as unlabeled
y_limited = np.copy(y)
y_limited[unlabeled_indices] = -1  # Assign -1 to unlabeled points

# Step 3: Train Label Propagation Model
label_prop_model = LabelPropagation(kernel="knn", n_neighbors=5)  # k-NN based propagation
label_prop_model.fit(X, y_limited)

# Step 4: Predict Labels for All Data Points
y_pred = label_prop_model.predict(X)

# Step 5: Plot Original Data (Before Label Propagation)
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c='gray', edgecolors='k', alpha=0.3, label="Unlabeled")
labeled_idx = np.where(y_limited != -1)[0]  # Find labeled points
plt.scatter(X[labeled_idx, 0], X[labeled_idx, 1], c=y[labeled_idx], cmap='coolwarm', edgecolors='k', s=100, label="Labeled")
plt.title("Original Data (Labeled + Unlabeled)")
plt.xlabel("X1")
plt.ylabel("X2")
plt.legend()

# Step 6: Plot Predicted Labels (After Label Propagation)
plt.subplot(1, 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='coolwarm', edgecolors='k', alpha=0.7)
plt.title("Predicted Labels After Label Propagation")
plt.xlabel("X1")
plt.ylabel("X2")

plt.show()

```
