```
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_moons
from sklearn.neural_network import MLPClassifier # For a simple supervised baseline
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from matplotlib.colors import ListedColormap

# 1. Generate synthetic data (e.g., two interleaving half-moons)
n_samples = 500
X, y = make_moons(n_samples=n_samples, noise=0.15, random_state=42)

# Scale features for neural networks
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Select a small subset as labeled data
n_labeled_per_class = 10 # Only 10 labeled points per class
labeled_indices = []
for i in np.unique(y):
    class_indices = np.where(y == i)[0]
    labeled_indices.extend(np.random.choice(class_indices, n_labeled_per_class, replace=False))

unlabeled_indices = np.setdiff1d(np.arange(n_samples), labeled_indices)

X_labeled = X_scaled[labeled_indices]
y_labeled = y[labeled_indices]
X_unlabeled = X_scaled[unlabeled_indices]


# Create a meshgrid for plotting decision boundaries
x_min, x_max = X_scaled[:, 0].min() - .5, X_scaled[:, 0].max() + .5
y_min, y_max = X_scaled[:, 1].min() - .5, X_scaled[:, 1].max() + .5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 100),
                     np.linspace(y_min, y_max, 100))

# Create color maps
cmap_light = ListedColormap(['#FFCCCC', '#CCE5FF']) # Light colors for decision regions
cmap_bold = ListedColormap(['#FF0000', '#0000FF'])   # Bold colors for classes

plt.figure(figsize=(14, 7))

# --- Subplot 1: Purely Supervised Learning (Baseline) ---
ax1 = plt.subplot(1, 2, 1)
ax1.set_title("Supervised Learning (Limited Labeled Data)")

# Train a simple MLP on only the labeled data
# This represents a purely supervised model that doesn't leverage unlabeled data
supervised_model = MLPClassifier(hidden_layer_sizes=(10,), max_iter=1000, random_state=42, activation='relu', solver='adam')
supervised_model.fit(X_labeled, y_labeled)

# Plot the decision boundary
Z_supervised = supervised_model.predict_proba(np.c_[xx.ravel(), yy.ravel()])[:, 1]
Z_supervised = Z_supervised.reshape(xx.shape)
ax1.contourf(xx, yy, Z_supervised, cmap=cmap_light, alpha=0.8)

# Plot labeled data
ax1.scatter(X_labeled[:, 0], X_labeled[:, 1], c=y_labeled, cmap=cmap_bold,
            edgecolor='k', s=100, label='Labeled Data')
# Plot unlabeled data (as light gray)
ax1.scatter(X_unlabeled[:, 0], X_unlabeled[:, 1], c='lightgray', alpha=0.6, s=30, label='Unlabeled Data')

ax1.legend()
ax1.set_xlim(xx.min(), xx.max())
ax1.set_ylim(yy.min(), yy.max())
ax1.set_xlabel('Feature 1')
ax1.set_ylabel('Feature 2')
ax1.text(0.05, 0.95, "Poor Decision Boundary\n(ignoring unlabeled data structure)",
         transform=ax1.transAxes, fontsize=10, verticalalignment='top', bbox=dict(boxstyle='round,pad=0.5', fc='white', ec='none', alpha=0.8))


# --- Subplot 2: Intrinsically Semi-Supervised Deep Learning (Conceptual) ---
ax2 = plt.subplot(1, 2, 2)
ax2.set_title("Intrinsically Semi-Supervised Deep Learning (Conceptual)")

# For conceptual visualization, we'll draw an ideal boundary that a semi-supervised model would learn.
# In a real scenario, this would be learned by a sophisticated model using unlabeled data.
# This represents a boundary that respects the underlying data manifold.
Z_semisup_conceptual = (xx > yy).astype(float) # A simple, illustrative non-linear boundary
Z_semisup_conceptual = (X[:,0] > 0.5 * (X[:,1] + X[:,1]**2)).astype(float) # More complex for make_moons
# For make_moons, a good conceptual boundary follows the "S" shape
Z_conceptual_boundary = np.zeros(xx.shape)
# Approximate the moon boundary
# You would get this from a trained semi-supervised model
for i in range(xx.shape[0]):
    for j in range(xx.shape[1]):
        # Simple conceptual classification based on y-coordinate relative to x-coordinate adjusted
        # This is just for demonstration, not an actual learned boundary
        if yy[i,j] > 0.5 * (xx[i,j]**2) - 0.1: # Approximate moon shape
             Z_conceptual_boundary[i,j] = 1
        else:
             Z_conceptual_boundary[i,j] = 0

# A slightly more refined conceptual boundary for make_moons
# Based on the approximate separation line in the middle of the "gap"
# This is manually crafted for illustration, not from a model.
distance_to_center_moon1 = np.sqrt((xx - 0.0)**2 + (yy - 0.5)**2)
distance_to_center_moon2 = np.sqrt((xx - 1.0)**2 + (yy - 0.0)**2)
Z_semisup_conceptual = (distance_to_center_moon1 < distance_to_center_moon2).astype(float)


ax2.contourf(xx, yy, Z_semisup_conceptual, cmap=cmap_light, alpha=0.8)

# Plot labeled data
ax2.scatter(X_labeled[:, 0], X_labeled[:, 1], c=y_labeled, cmap=cmap_bold,
            edgecolor='k', s=100, label='Labeled Data')
# Plot unlabeled data (as light gray)
ax2.scatter(X_unlabeled[:, 0], X_unlabeled[:, 1], c='lightgray', alpha=0.6, s=30, label='Unlabeled Data')

ax2.legend()
ax2.set_xlim(xx.min(), xx.max())
ax2.set_ylim(yy.min(), yy.max())
ax2.set_xlabel('Feature 1')
ax2.set_ylabel('Feature 2')
ax2.text(0.05, 0.95, "Improved Decision Boundary\n(leveraging unlabeled data manifold)",
         transform=ax2.transAxes, fontsize=10, verticalalignment='top', bbox=dict(boxstyle='round,pad=0.5', fc='white', ec='none', alpha=0.8))


plt.tight_layout()
plt.show()
```
