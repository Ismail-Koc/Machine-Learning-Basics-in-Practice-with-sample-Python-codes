```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Step 1: Generate a Non-Linearly Separable Dataset
X, y = make_moons(n_samples=300, noise=0.2, random_state=42)

# Step 2: Perform Unsupervised Feature Extraction using PCA
pca = PCA(n_components=2)  # Reduce to 2 principal components
X_pca = pca.fit_transform(X)

# Step 3: Train a Supervised Model (SVM) on the Extracted Features
X_train, X_test, y_train, y_test = train_test_split(X_pca, y, test_size=0.3, random_state=42)
model = SVC(kernel='linear', C=1)
model.fit(X_train, y_train)

# Step 4: Make Predictions
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

# Step 5: Plot Original Data and Extracted Feature Space
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Original Data
axes[0].scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='k', alpha=0.7)
axes[0].set_title("Original Data (Raw Features)")

# Extracted Feature Space
axes[1].scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='coolwarm', edgecolors='k', alpha=0.7)
axes[1].set_title("Feature Space After PCA")

plt.show()

# Step 6: Print Model Performance
print(f"Accuracy after feature extraction: {accuracy:.2f}")
```
