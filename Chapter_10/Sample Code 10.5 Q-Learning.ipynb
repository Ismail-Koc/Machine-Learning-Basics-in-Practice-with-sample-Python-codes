```
import numpy as np
import matplotlib.pyplot as plt
import random

# Define Grid World parameters
GRID_SIZE = 4
GAMMA = 0.9  # Discount factor
ALPHA = 0.1  # Learning rate
EPSILON = 0.2  # Exploration rate
EPISODES = 1000  # Training episodes

ACTIONS = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # Up, Down, Left, Right
ACTION_NAMES = ["Up", "Down", "Left", "Right"]

# Initialize Q-table
Q = np.zeros((GRID_SIZE, GRID_SIZE, len(ACTIONS)))

# Function to check if a state is terminal
def is_terminal(state):
    return state == (GRID_SIZE-1, GRID_SIZE-1)

# Choose an action using epsilon-greedy strategy
def choose_action(state):
    if random.uniform(0, 1) < EPSILON:
        return random.randint(0, len(ACTIONS) - 1)  # Explore
    else:
        return np.argmax(Q[state[0], state[1]])  # Exploit

# Training loop for Q-learning
for episode in range(EPISODES):
    state = (0, 0)  # Start at top-left
    while not is_terminal(state):
        action_index = choose_action(state)
        action = ACTIONS[action_index]

        # Determine next state
        ni, nj = state[0] + action[0], state[1] + action[1]
        if 0 <= ni < GRID_SIZE and 0 <= nj < GRID_SIZE:
            next_state = (ni, nj)
        else:
            next_state = state  # Stay in place if out of bounds

        # Get reward
        reward = -1 if not is_terminal(next_state) else 0

        # Q-learning update rule
        Q[state[0], state[1], action_index] += ALPHA * (
            reward + GAMMA * np.max(Q[next_state[0], next_state[1]]) - Q[state[0], state[1], action_index]
        )

        state = next_state  # Move to next state

# Extract optimal policy
policy = np.array([[ACTION_NAMES[np.argmax(Q[i, j])] if not is_terminal((i, j)) else "G"
                    for j in range(GRID_SIZE)] for i in range(GRID_SIZE)])

# Print the learned policy
print("\nOptimal Policy (G = Goal):\n")
for row in policy:
    print(row)

# Plot Q-values
plt.figure(figsize=(6, 6))
plt.imshow(np.max(Q, axis=2), cmap="coolwarm", interpolation="nearest")
plt.colorbar(label="Max Q-value")
plt.title("Learned Q-values (Max per state)")
plt.show()
```
