```
import numpy as np
import matplotlib.pyplot as plt

# Define Grid World parameters
GRID_SIZE = 3
GAMMA = 0.9  # Discount factor
THRESHOLD = 1e-4  # Convergence threshold
ACTIONS = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # Up, Down, Left, Right
ACTION_NAMES = ["Up", "Down", "Left", "Right"]
ACTION_REWARDS = [-2, -1, -1.5, -1]  # Different rewards per action

# Initialize rewards
REWARDS = np.full((GRID_SIZE, GRID_SIZE), -1)  # Default reward
REWARDS[GRID_SIZE-1, GRID_SIZE-1] = 0  # Goal state reward

# Initialize value function
V = np.zeros((GRID_SIZE, GRID_SIZE))

# Function to check if a state is terminal
def is_terminal(state):
    return state == (GRID_SIZE-1, GRID_SIZE-1)

# Function to plot the value function at each step
def plot_value_function(V, iteration):
    plt.figure(figsize=(6, 6))
    plt.imshow(V, cmap='coolwarm', interpolation='nearest')
    plt.colorbar(label="Value")
    plt.title(f"Value Function at Iteration {iteration}")
    plt.show()

# Value Iteration using Bellman Equation
iteration = 0
while True:
    delta = 0  # To track convergence
    new_V = np.copy(V)

    print(f"\nIteration {iteration} - Bellman Equation at every step:\n")

    for i in range(GRID_SIZE):
        for j in range(GRID_SIZE):
            if is_terminal((i, j)):  # Skip terminal state
                continue

            value_list = []
            bellman_steps = []  # To store equations for printing

            for action_index, action in enumerate(ACTIONS):
                ni, nj = i + action[0], j + action[1]

                if 0 <= ni < GRID_SIZE and 0 <= nj < GRID_SIZE:  # Valid move
                    next_state_value = V[ni, nj]
                else:  # If out of bounds, remain in the same state
                    next_state_value = V[i, j]

                # Apply Bellman equation with different action rewards
                action_value = ACTION_REWARDS[action_index] + GAMMA * next_state_value
                value_list.append(action_value)

                # Store the Bellman equation as a string for this step
                equation = f"V({i},{j}) = {ACTION_REWARDS[action_index]} + {GAMMA} * V({ni},{nj}) = {action_value:.2f}"
                bellman_steps.append((ACTION_NAMES[action_index], equation))

            # Print the Bellman equation calculations for this state
            print(f"State ({i},{j}):")
            for action_name, eq in bellman_steps:
                print(f"  {action_name}: {eq}")

            # Update value function with max value
            new_V[i, j] = max(value_list)
            delta = max(delta, abs(new_V[i, j] - V[i, j]))  # Track max change

    V = new_V
    iteration += 1

    # Plot after every iteration
    plot_value_function(V, iteration)

    if delta < THRESHOLD:  # Convergence check
        break

print(f"Converged in {iteration} iterations.\nFinal Optimal Value Function:")
print(V)
```
