```
import numpy as np

# Step 1: Define the Gridworld Environment
# A 4x4 grid with states numbered 0-15
grid_size = 4
num_states = grid_size * grid_size
terminal_states = [0, 15]  # Terminal states with rewards
rewards = {0: 0, 15: 1}  # Reward for reaching terminal states
gamma = 0.9  # Discount factor

# Step 2: Define the Environment Dynamics
def step(state, action):
    """Execute an action and return the next state and reward."""
    row, col = divmod(state, grid_size)
    if state in terminal_states:
        return state, rewards[state]

    # Actions: 0 = Up, 1 = Right, 2 = Down, 3 = Left
    if action == 0:  # Up
        row = max(0, row - 1)
    elif action == 1:  # Right
        col = min(grid_size - 1, col + 1)
    elif action == 2:  # Down
        row = min(grid_size - 1, row + 1)
    elif action == 3:  # Left
        col = max(0, col - 1)

    next_state = row * grid_size + col
    reward = rewards.get(next_state, 0)
    return next_state, reward

# Step 3: TD(0) Learning
def td_learning(num_episodes, alpha=0.1):
    """Estimate state values using TD(0) learning."""
    V = np.zeros(num_states)  # Initialize state-value function
    for _ in range(num_episodes):
        state = np.random.choice(range(num_states))  # Random initial state

        while state not in terminal_states:
            action = np.random.choice([0, 1, 2, 3])  # Random action (exploration)
            next_state, reward = step(state, action)

            # TD(0) Update Rule
            td_error = reward + gamma * V[next_state] - V[state]
            V[state] += alpha * td_error

            state = next_state  # Move to the next state
    return V

# Step 4: Run TD Learning
num_episodes = 1000
V = td_learning(num_episodes)
print("Estimated State-Value Function:")
print(V.reshape(grid_size, grid_size))
```
