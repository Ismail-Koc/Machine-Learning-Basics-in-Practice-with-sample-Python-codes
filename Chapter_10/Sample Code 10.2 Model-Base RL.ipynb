```
import numpy as np
import random

# Step 1: Define GridWorld Environment
class GridWorld:
    def __init__(self, size=5):
        self.size = size
        self.state = (0, 0)  # Start state
        self.goal = (size-1, size-1)  # Goal position

    def step(self, action):
        """Take action and return (next_state, reward)."""
        x, y = self.state
        if action == "UP":
            x = max(0, x - 1)
        elif action == "DOWN":
            x = min(self.size - 1, x + 1)
        elif action == "LEFT":
            y = max(0, y - 1)
        elif action == "RIGHT":
            y = min(self.size - 1, y + 1)

        self.state = (x, y)
        reward = 1 if self.state == self.goal else -0.1  # Reward at the goal
        return self.state, reward

    def reset(self):
        """Reset environment."""
        self.state = (0, 0)
        return self.state

# Step 2: Define Model-Based Reinforcement Learning (Dyna-Q)
class DynaQAgent:
    def __init__(self, env, alpha=0.1, gamma=0.9, epsilon=0.1, planning_steps=5):
        self.env = env
        self.q_table = {}  # Q-values
        self.model = {}  # Learned transition model
        self.alpha = alpha  # Learning rate
        self.gamma = gamma  # Discount factor
        self.epsilon = epsilon  # Exploration rate
        self.planning_steps = planning_steps  # Number of simulated updates

    def choose_action(self, state):
        """Epsilon-greedy action selection."""
        if random.random() < self.epsilon:  # Explore
            return random.choice(["UP", "DOWN", "LEFT", "RIGHT"])
        return max(self.q_table.get(state, {}), key=self.q_table.get(state, {}).get, default="UP")  # Exploit

    def update_q_table(self, state, action, next_state, reward):
        """Q-learning update."""
        if state not in self.q_table:
            self.q_table[state] = {a: 0 for a in ["UP", "DOWN", "LEFT", "RIGHT"]}

        max_next_q = max(self.q_table.get(next_state, {}).values(), default=0)
        self.q_table[state][action] += self.alpha * (reward + self.gamma * max_next_q - self.q_table[state][action])

    def update_model(self, state, action, next_state, reward):
        """Store transitions in the model."""
        self.model[(state, action)] = (next_state, reward)

    def plan(self):
        """Simulate experiences using learned model."""
        for _ in range(self.planning_steps):
            state, action = random.choice(list(self.model.keys()))
            next_state, reward = self.model[(state, action)]
            self.update_q_table(state, action, next_state, reward)  # Simulated learning

    def train(self, episodes=100):
        """Train the agent using real interactions + model-based planning."""
        for episode in range(episodes):
            state = self.env.reset()
            while state != self.env.goal:
                action = self.choose_action(state)
                next_state, reward = self.env.step(action)
                self.update_q_table(state, action, next_state, reward)
                self.update_model(state, action, next_state, reward)
                self.plan()  # Perform simulated updates
                state = next_state
#            print(f"Episode {episode+1} completed!")

# Step 3: Train Agent with Model-Based RL
env = GridWorld(size=5)
agent = DynaQAgent(env)
agent.train(episodes=20)

# Display learned Q-values
print("\nLearned Q-values after Model-Based Reinforcement Learning:")
for state, actions in agent.q_table.items():
    print(state, actions)
```
