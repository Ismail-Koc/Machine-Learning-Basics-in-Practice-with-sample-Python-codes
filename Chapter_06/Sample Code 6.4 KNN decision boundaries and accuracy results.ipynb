```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from matplotlib.colors import ListedColormap

# Generate random data with 3 distinct clusters
n_samples = 50
centers = [[2, 2], [7, 7], [3, 8]]
X, y = make_blobs(n_samples=n_samples, centers=centers, random_state=42, cluster_std=1.0)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Define the k values for the decision boundary plots
k_values_boundary = [1, 3, 5, 10, 20, 30]

# Define a range of k values to test for accuracy
k_values_accuracy = range(1, int(0.7 * n_samples))  # Up to around sqrt(n) is often reasonable
accuracies = []

# Calculate accuracies for different k values
for k in k_values_accuracy:
    knn_accuracy = KNeighborsClassifier(n_neighbors=k)
    knn_accuracy.fit(X_train, y_train)
    y_pred = knn_accuracy.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracies.append(accuracy)

# Create a colormap for the classes
cmap_light = ListedColormap(['#FFAAAA', '#AAFFAA', '#AAAAFF'])
cmap_bold = ListedColormap(['#FF0000', '#00FF00', '#0000FF'])

# Create the figure with subplots
fig = plt.figure(figsize=(18, 12))
gs = fig.add_gridspec(3, 3)  # 3 rows, 3 columns for better layout

# Plot decision boundaries for different k values
for i, k in enumerate(k_values_boundary):
    ax_boundary = fig.add_subplot(gs[int(i/3), i % 3])
    knn_boundary = KNeighborsClassifier(n_neighbors=k)
    knn_boundary.fit(X_train, y_train)

    # Create a meshgrid for plotting decision boundaries
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                         np.arange(y_min, y_max, 0.02))
    Z = knn_boundary.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    ax_boundary.contourf(xx, yy, Z, cmap=cmap_light)

    # Plot the training points
    scatter_boundary = ax_boundary.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap=cmap_bold, edgecolor='k', s=40, label='Train')
    ax_boundary.scatter(X_test[:, 0], X_test[:, 1], c=y_test, cmap=cmap_bold, edgecolor='gray', marker='o', s=60, label='Test')

    ax_boundary.set_title(f'KNN (k={k}) Decision Boundary')
    ax_boundary.set_xlabel('Feature 1')
    ax_boundary.set_ylabel('Feature 2')
    if i == 0:
        ax_boundary.legend()

# Plot the k-accuracy graph
ax_accuracy = fig.add_subplot(gs[2, :])
ax_accuracy.plot(k_values_accuracy, accuracies, marker='o', linestyle='-')
ax_accuracy.set_title('KNN Accuracy vs. K Value')
ax_accuracy.set_xlabel('Number of Neighbors (K)')
ax_accuracy.set_ylabel('Accuracy on Test Set')
ax_accuracy.set_xticks(list(k_values_accuracy))
ax_accuracy.grid(True)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.suptitle('KNN Decision Boundaries and K-Accuracy Graph (50 Random Data Samples)', fontsize=16)
plt.show()
```
