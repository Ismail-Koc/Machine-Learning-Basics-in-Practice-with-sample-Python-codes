```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Generate synthetic data
np.random.seed(42)
X = np.linspace(0, 10, 100).reshape(-1, 1)  # Independent variable
true_beta_0 = 5  # True intercept (β0)
true_beta_1 = 3  # True slope (β1)
epsilon = np.random.normal(scale=2, size=len(X))  # Random error (𝜖)
Y = true_beta_0 + true_beta_1 * X.flatten() + epsilon  # Dependent variable (Y)

# Fit a linear regression model
model = LinearRegression()
model.fit(X, Y)
Y_pred = model.predict(X)

# Retrieve the fitted coefficients
fitted_beta_0 = model.intercept_
fitted_beta_1 = model.coef_[0]

# Calculate the Mean Squared Error (MSE)
mse = mean_squared_error(Y, Y_pred)

# Plot the data, true line, and fitted regression line
plt.figure(figsize=(10, 6))

# Scatter plot of data points
plt.scatter(X, Y, color='orange’, label='Observed Data’, alpha=0.7)

# True regression line (Y = β0 + β1X)
plt.plot(X, true_beta_0 + true_beta_1 * X, color='green’, linestyle='--’, label='True Line (Y = β0 + β1X)’, linewidth=2)

# Fitted regression line
plt.plot(X, Y_pred, color='blue’, label=f’Fitted Line (Y = {fitted_beta_0:.2f} + {fitted_beta_1:.2f}X)’, linewidth=2)

# Titles and labels
plt.title('Linear Regression: Y = β0 + β1X + 𝜖’, fontsize=16)
plt.xlabel('X (Independent Variable)’, fontsize=12)
plt.ylabel('Y (Dependent Variable)’, fontsize=12)
plt.legend(fontsize=10)
plt.grid(alpha=0.5)

# Show the plot
plt.show()

# Print the coefficients and MSE
print(f"True Coefficients: β0 = {true_beta_0}, β1 = {true_beta_1}")
print(f"Fitted Coefficients: β0 = {fitted_beta_0:.2f}, β1 = {fitted_beta_1:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
```
