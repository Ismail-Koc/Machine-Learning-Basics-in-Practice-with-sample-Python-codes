```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.datasets import make_blobs

# Generate sample data with two classes (non-linearly separable for sigmoid to be interesting)
X, y = make_blobs(n_samples=100, centers=[[2, 2], [7, 7]], random_state=42, cluster_std=2.0)

# Create an SVM classifier with a sigmoid kernel
clf = SVC(kernel='sigmoid', C=1.0, gamma='auto')  # gamma='auto' for automatic calculation
clf.fit(X, y)

# Create a meshgrid for plotting decision boundaries
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                     np.linspace(y_min, y_max, 200))
Z = clf.decision_function(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot the data points
plt.figure(figsize=(10, 8))
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap=plt.cm.RdYlBu)

# Plot the decision boundary
plt.contour(xx, yy, Z, colors='k', levels=[0], linestyles=['-'])

# Highlight support vectors (sigmoid kernel doesn't have a simple margin like linear)
support_vectors = clf.support_vectors_
plt.scatter(support_vectors[:, 0], support_vectors[:, 1], s=100, facecolors='none', edgecolor='k', label='Support Vectors')
plt.legend()

plt.title('Sample Support Vector Machine with Two Classes (Sigmoid Kernel)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True)
plt.show()
```
