```
import gensim
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk

# Download NLTK resources
nltk.download("punkt")

# Step 1: Sample Corpus
corpus = [
    "Natural Language Processing is fascinating.",
    "Word embeddings are helpful for machine learning tasks.",
    "Word2Vec generates vector representations of words.",
    "Understanding semantics through embeddings is important.",
    "Machine learning models leverage Word2Vec embeddings effectively."
]

# Step 2: Tokenize Sentences
tokenized_corpus = [word_tokenize(sentence.lower()) for sentence in corpus]
print("Tokenized Corpus:")
print(tokenized_corpus)

# Step 3: Train Word2Vec Model
# Parameters:
# - vector_size: Dimension of the embedding vectors (e.g., 100)
# - window: Context window size for CBOW/Skip-Gram (e.g., 5)
# - min_count: Minimum word frequency to include in training
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,  # Use CBOW (sg=1 for Skip-Gram)
    epochs=10
)

# Step 4: Explore the Model
# Get the embedding for a word
word = "word2vec"
if word in model.wv:
    print(f"\nEmbedding for '{word}':")
    print(model.wv[word])

# Find most similar words
similar_words = model.wv.most_similar("word2vec", topn=5)
print(f"\nWords similar to '{word}':")
for sim_word, score in similar_words:
    print(f"{sim_word}: {score}")

# Solve analogies
# Example: king - man + woman ≈ queen (if corpus permits such relations)
try:
    analogy = model.wv.most_similar(positive=["machine", "learning"], negative=["processing"])
    print("\nAnalogy Result:")
    print(analogy)
except KeyError:
    print("\nAnalogy example cannot be solved due to limited corpus vocabulary.")

# Save and Load the Model
model.save("word2vec_sample.model")
loaded_model = Word2Vec.load("word2vec_sample.model")
print("\nModel saved and reloaded successfully.")
```
