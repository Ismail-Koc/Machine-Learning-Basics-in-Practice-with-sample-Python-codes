```
from transformers import GPT2Tokenizer, GPT2LMHeadModel

# Step 1: Load Pretrained GPT Model and Tokenizer
model_name = "gpt2"  # Pretrained GPT-2 model from Hugging Face
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

# Step 2: Input Text
input_text = "Natural Language Processing is fascinating because"

# Step 3: Tokenize Input
input_ids = tokenizer.encode(input_text, return_tensors="pt")  # Convert text to token IDs

# Step 4: Generate Text
output = model.generate(
    input_ids,
    max_length=50,  # Maximum length of the generated text
    num_beams=5,    # Beam search for more coherent output
    no_repeat_ngram_size=2,  # Avoid repeating phrases
    early_stopping=True      # Stop generating early if conditions are met
)

# Step 5: Decode Generated Text
generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
print("Generated Text:")
print(generated_text)
```
