```
from transformers import BertTokenizer, BertModel
import torch

# Load pre-trained BERT tokenizer and model
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')

# Example sentence
sentence = "The quick brown fox jumps over the lazy dog."

# Tokenize the sentence
inputs = tokenizer(sentence, return_tensors="pt")

# Get the model's output
with torch.no_grad():
    outputs = model(**inputs)

# The `outputs` object contains various information, including:
# - `last_hidden_state`: The sequence of hidden states at the output of the last layer.
#                        Shape: (batch_size, sequence_length, hidden_size)
# - `pooler_output`: The output of the [CLS] token, passed through a linear layer and tanh activation.
#                    Shape: (batch_size, hidden_size)

# Extract the last hidden state (contextualized word embeddings)
last_hidden_states = outputs.last_hidden_state
print("Shape of last hidden states:", last_hidden_states.shape)
# Shape: torch.Size([1, 12, 768])
# - 1: batch size (we processed one sentence)
# - 12: sequence length (number of tokens, including special tokens)
# - 768: hidden size (dimensionality of the BERT embeddings for 'bert-base-uncased')

# Get the embedding for the word "fox"
# Find the index of "fox" in the tokenized input
tokens = tokenizer.tokenize(sentence)
fox_index = tokens.index('fox') + 1  # +1 because of the [CLS] token at the beginning

embedding_fox = last_hidden_states[0, fox_index, :]
print("\nEmbedding for 'fox':")
print(embedding_fox[:10]) # Print only the first 10 dimensions

# Extract the pooled output (sentence-level representation)
pooler_output = outputs.pooler_output
print("\nShape of pooler output:", pooler_output.shape)
print("Pooled output (sentence embedding):")
print(pooler_output[:10])

# You can now use these embeddings for downstream tasks.
# For example, you could:
# - Fine-tune the model on a classification task using the `pooler_output`.
# - Use the `last_hidden_states` for token-level tasks like Named Entity Recognition.

# To use BERT for a specific task, you would typically use task-specific heads
# provided by the Transformers library (e.g., BertForSequenceClassification,
# BertForTokenClassification, BertForQuestionAnswering).
```
