```
import fasttext
import fasttext.util

# Step 1: Prepare a Sample Corpus
# Save the corpus to a text file for training FastText
corpus = [
    "Natural Language Processing is fascinating.",
    "Word embeddings are helpful for machine learning tasks.",
    "Word2Vec generates vector representations of words.",
    "Understanding semantics through embeddings is important.",
    "Machine learning models leverage Word2Vec embeddings effectively."
]
corpus_file = "sample_corpus.txt"
with open(corpus_file, "w") as f:
    f.write("\n".join(corpus))

# Step 2: Train FastText Embeddings
# Train a Skip-Gram model using FastText
model = fasttext.train_unsupervised(corpus_file, model="skipgram", dim=100, minCount=1)

# Step 3: Explore Word Embeddings
# Get the embedding for a specific word
word = "machine"
if word in model.words:
    print(f"\nEmbedding for '{word}':\n", model[word])
else:
    print(f"'{word}' not found in FastText embeddings.")

# Step 4: Find Most Similar Words
similar_words = model.get_nearest_neighbors("machine", k=5)
print(f"\nWords similar to '{word}':")
for score, similar_word in similar_words:
    print(f"{similar_word}: {score:.4f}")

# Step 5: Solve Analogies
# Example analogy: king - man + woman ≈ queen
try:
    analogy = model.get_analogies("king", "man", "woman", k=5)
    print("\nAnalogy Example (king - man + woman ≈ ?):")
    for score, word in analogy:
        print(f"{word}: {score:.4f}")
except Exception as e:
    print("\nError solving analogy:", str(e))
```
