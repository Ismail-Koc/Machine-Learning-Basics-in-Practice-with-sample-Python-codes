```
import numpy as np
from sklearn.preprocessing import RobustScaler

# Step 1: Sample Dataset
data = np.array([[1], [2], [3], [4], [100]])  # Example dataset with an outlier

# Step 2: Apply Robust Scaler
scaler = RobustScaler()
scaled_data = scaler.fit_transform(data)

print("Original Data:")
print(data)
print("\nScaled Data:")
print(scaled_data)
```
