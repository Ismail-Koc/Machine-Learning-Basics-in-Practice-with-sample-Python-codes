```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
from sklearn.datasets import make_classification

# Generate synthetic dataset
X, y = make_classification(n_samples=200, n_features=5, random_state=42)

# Split into training & testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Overfit model: Deep decision tree (no restrictions)
overfit_tree = DecisionTreeClassifier(max_depth=None, random_state=42)
overfit_tree.fit(X_train, y_train)

# Pruned model: Limit depth to prevent overfitting
pruned_tree = DecisionTreeClassifier(max_depth=3, random_state=42)  # Restrict depth to generalize better
pruned_tree.fit(X_train, y_train)

# Predictions
y_pred_overfit = overfit_tree.predict(X_test)
y_pred_pruned = pruned_tree.predict(X_test)

# Evaluate accuracy
train_acc_overfit = accuracy_score(y_train, overfit_tree.predict(X_train))
test_acc_overfit = accuracy_score(y_test, y_pred_overfit)

train_acc_pruned = accuracy_score(y_train, pruned_tree.predict(X_train))
test_acc_pruned = accuracy_score(y_test, y_pred_pruned)

print("Overfitted Model - Train Accuracy:", train_acc_overfit)
print("Overfitted Model - Test Accuracy:", test_acc_overfit)

print("\nPruned Model - Train Accuracy:", train_acc_pruned)
print("Pruned Model - Test Accuracy:", test_acc_pruned)

# Visualize the trees
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plot_tree(overfit_tree, filled=True)
plt.title("Overfitted Decision Tree")

plt.subplot(1, 2, 2)
plot_tree(pruned_tree, filled=True)
plt.title("Pruned Decision Tree")

plt.show()
```
