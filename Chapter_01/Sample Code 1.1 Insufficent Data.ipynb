```
from imblearn.over_sampling import SMOTE
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# Sample dataset (features)
X = np.array([[1, 10], [2, 20], [3, 30], [4, 40], [5, 50]])
y = np.array([0, 0, 0, 1, 1])  # Binary class labels

# Convert to DataFrame for readability
df_original = pd.DataFrame(X, columns=['Feature1', 'Feature2'])
df_original['Target'] = y

# Display the original dataset
print("Original Data:")
print(df_original)

# Apply SMOTE for synthetic data generation
smote = SMOTE(sampling_strategy='auto', random_state=42, k_neighbors=1)
X_resampled, y_resampled = smote.fit_resample(X, y)

# Convert to DataFrame for readability
df_resampled = pd.DataFrame(X_resampled, columns=['Feature1', 'Feature2'])
df_resampled['Target'] = y_resampled

# Display the new dataset
print("Resampled Data:")
print(df_resampled)
```
