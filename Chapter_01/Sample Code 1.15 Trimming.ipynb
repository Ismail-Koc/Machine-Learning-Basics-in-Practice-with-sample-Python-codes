```
import numpy as np
import pandas as pd

# Step 1: Sample Dataset
data = [1, 2, 3, 4, 5, 6, 7, 100, 120]  # Example dataset with outliers
df = pd.DataFrame(data, columns=["Value"])

# Step 2: Calculate IQR and Define Bounds
Q1 = df["Value"].quantile(0.25)  # First quartile (25th percentile)
Q3 = df["Value"].quantile(0.75)  # Third quartile (75th percentile)
IQR = Q3 - Q1  # Interquartile range

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")

# Step 3: Trim Outliers
trimmed_data = df[(df["Value"] >= lower_bound) & (df["Value"] <= upper_bound)]

print("\nTrimmed Dataset:")
print(trimmed_data)
```
