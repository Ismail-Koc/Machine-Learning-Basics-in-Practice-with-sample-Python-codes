```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error

# Generate non-linear synthetic data
np.random.seed(42)
X = np.linspace(0, 10, 30).reshape(-1, 1)
y = np.sin(X).ravel() + np.random.normal(scale=0.2, size=X.shape[0])  # Adding noise

# Train an underfitting model: Linear Regression
linear_model = LinearRegression()
linear_model.fit(X, y)
y_pred_linear = linear_model.predict(X)

# Train a better model: Polynomial Regression (Degree 5)
poly_model = make_pipeline(PolynomialFeatures(degree=5), LinearRegression())
poly_model.fit(X, y)
y_pred_poly = poly_model.predict(X)

# Evaluate Performance
mse_linear = mean_squared_error(y, y_pred_linear)
mse_poly = mean_squared_error(y, y_pred_poly)

print("Underfitting Model (Linear Regression) - MSE:", mse_linear)
print("Improved Model (Polynomial Regression) - MSE:", mse_poly)

# Plot Results
plt.scatter(X, y, label="Actual Data", color="blue")
plt.plot(X, y_pred_linear, label="Underfitted Model (Linear)", linestyle="dashed", color="red")
plt.plot(X, y_pred_poly, label="Improved Model (Polynomial)", linestyle="solid", color="green")

plt.xlabel("Feature (X)")
plt.ylabel("Target (y)")
plt.legend()
plt.title("Underfitting vs. Improved Model")
plt.show()
```
