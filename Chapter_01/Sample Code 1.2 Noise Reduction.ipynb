```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Original dataset with missing values and outliers
data = {'Feature1': [10, 20, np.nan, 40, 50, 1000],  # Contains a missing value and an outlier (1000)
        'Feature2': [5, 15, 25, np.nan, 45, 55],  # Contains a missing value
        'Target': [1, 0, 1, 1, 0, 1]}

df_original = pd.DataFrame(data)
print("Original Data:\n", df_original)

# Handling missing values: Fill with mean
df_cleaned = df_original.copy()
df_cleaned.fillna(df_cleaned.mean(), inplace=True)

# Handling outliers: Replace extreme values with median using Interquartile Range (IQR)
Q1 = df_cleaned['Feature1'].quantile(0.25)
Q3 = df_cleaned['Feature1'].quantile(0.75)
IQR = Q3 - Q1
outlier_threshold_low = Q1 - 1.5 * IQR
outlier_threshold_high = Q3 + 1.5 * IQR

df_cleaned['Feature1'] = np.where((df_cleaned['Feature1'] > outlier_threshold_high) | (df_cleaned['Feature1'] < outlier_threshold_low),
                                  df_cleaned['Feature1'].median(), df_cleaned['Feature1'])

print("\nCleaned Data:\n", df_cleaned)

# Visualizing the difference with boxplots
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.boxplot(df_original.dropna()['Feature1'])
plt.title("Original Data (Feature1)")

plt.subplot(1, 2, 2)
plt.boxplot(df_cleaned['Feature1'])
plt.title("Cleaned Data (Feature1)")

plt.show()
```
