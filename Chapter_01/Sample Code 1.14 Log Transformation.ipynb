```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Sample Dataset
data = [1, 5, 10, 20, 100, 200, 1000]  # Example dataset with outliers

# Convert to DataFrame
df = pd.DataFrame(data, columns=["Value"])

# Step 2: Apply Log Transformation
df["Log_Transformed"] = np.log1p(df["Value"])  # Use log1p (log(x + 1)) to handle zeros

# Step 3: Plot Original vs. Transformed Data
plt.figure(figsize=(8, 6))
plt.subplot(1, 2, 1)
plt.boxplot(df["Value"])
plt.title("Original Data")

plt.subplot(1, 2, 2)
plt.boxplot(df["Log_Transformed"])
plt.title("Log-Transformed Data")

plt.show()

# Step 4: Print Results
print("Original Data:")
print(df["Value"])
print("\nLog-Transformed Data:")
print(df["Log_Transformed"])
```
