```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from imblearn.over_sampling import SMOTE

# Step 1: Generate imbalanced dataset
X, y = make_classification(n_samples=1000, n_features=10, weights=[0.9, 0.1], random_state=42)

# Display class distribution before balancing
print("Original class distribution:", Counter(y))

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Step 2: Train model on imbalanced data
model_imbalanced = RandomForestClassifier(random_state=42)
model_imbalanced.fit(X_train, y_train)
y_pred_imbalanced = model_imbalanced.predict(X_test)

# Evaluate performance before fixing imbalance
print("\nPerformance on Imbalanced Data:")
print(classification_report(y_test, y_pred_imbalanced))

# Step 3: Apply SMOTE for balancing data
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

# Display class distribution after balancing
print("\nBalanced class distribution:", Counter(y_train_balanced))

# Step 4: Train model on SMOTE balanced data
model_balanced = RandomForestClassifier(random_state=42)
model_balanced.fit(X_train_balanced, y_train_balanced)
y_pred_balanced = model_balanced.predict(X_test)

# Evaluate performance after fixing imbalance
print("\nPerformance on Balanced Data (After SMOTE):")
print(classification_report(y_test, y_pred_balanced))

# Step 5: Visualization of class distribution
plt.figure(figsize=(8, 4))
plt.bar(["Original Minority", "Original Majority", "Balanced Minority", "Balanced Majority"],
        [Counter(y)[1], Counter(y)[0], Counter(y_train_balanced)[1], Counter(y_train_balanced)[0]],
        color=['red', 'blue', 'red', 'blue'])
plt.ylabel("Number of Samples")
plt.title("Class Distribution Before and After SMOTE")
plt.show()
```
