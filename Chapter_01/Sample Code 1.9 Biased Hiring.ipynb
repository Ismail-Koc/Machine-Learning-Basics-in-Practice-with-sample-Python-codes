```
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

# Sample biased hiring data (simplified example)
data = {
    'years_experience': [5, 8, 3, 6, 2, 7, 4, 9],
    'education_level': [3, 4, 2, 4, 1, 3, 2, 4],  # 1-4 scale
    'gender': ['male', 'female', 'female', 'male', 'female', 'male', 'female', 'male'],
    'hired': [1, 0, 0, 1, 0, 1, 0, 1]  # 1=hired, 0=not hired
}

df = pd.DataFrame(data)

# Convert gender to binary (problematic approach)
df['gender'] = df['gender'].map({'male': 1, 'female': 0})

# Features and target
X = df[['years_experience', 'education_level', 'gender']]
y = df['hired']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train model
model = RandomForestClassifier()
model.fit(X_train, y_train)

# Check feature importance
print("Feature importances:", model.feature_importances_)
```
