```
import numpy as np
import pandas as pd
from math import log2

# Sample Data
data = {
    ‘Customer’: [1, 2, 3, 4, 5, 6, 7, 8, 9],
    ‘Income Level’: [‘High’, ‘High’, ‘High’, ‘Medium’, ‘Medium’, ‘Medium’, ‘Low’, ‘Low’, ‘Low’],
    ‘Age Group’: [‘Young’, ‘Middle-aged’, ‘Senior’, ‘Young’, ‘Middle-aged’, ‘Senior’, ‘Young’, ‘Middle-aged’, ‘Senior’],
    ‘Buys Product (Yes/No)’: [‘No’, ‘No’, ‘Yes’, ‘Yes’, ‘No’, ‘Yes’, ‘Yes’, ‘Yes’, ‘No’]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Function to calculate entropy
def entropy(labels):
    label_counts = labels.value_counts()
    total = len(labels)
    entropy_val = 0
    for count in label_counts:
        p = count / total
        entropy_val -= p * log2(p)
    return entropy_val

# Function to calculate Information Gain
def information_gain(data, feature, target):
    # Calculate the entropy of the whole dataset (before the split)
    total_entropy = entropy(data[target])

    # Split the dataset by the feature
    feature_values = data[feature].unique()

    # Calculate the weighted entropy of the subsets
    weighted_entropy = 0
    for value in feature_values:
        subset = data[data[feature] == value]
        weighted_entropy += (len(subset) / len(data)) * entropy(subset[target])

    # Information Gain is the difference between total entropy and weighted entropy
    return total_entropy - weighted_entropy

# 1. Information Gain for the first layer (split by Income Level)
first_layer_ig = information_gain(df, ‘Income Level’, ‘Buys Product (Yes/No)’)
print(f"Information Gain for ‘Income Level’ (First Layer): {first_layer_ig:.4f}")

# 2. Information Gain for the second layer (split by Age Group, for High Income)
# Filter the data for High Income group
high_income_data = df[df[‘Income Level’] == ‘High’]

# Calculate Information Gain for Age Group within High Income group
second_layer_ig = information_gain(high_income_data, ‘Age Group’, ‘Buys Product (Yes/No)’)
print(f"Information Gain for ‘Age Group’ within High Income (Second Layer): {second_layer_ig:.4f}")
```
