```
import pandas as pd

# Sample Data
data = {
    'Customer': [1, 2, 3, 4, 5, 6, 7, 8, 9],
    'Income Level': ['High', 'High', 'High', 'Medium', 'Medium', 'Medium', 'Low', 'Low', 'Low'],
    'Buys Product (Yes/No)': ['Yes', 'No', 'Yes', 'Yes', 'No', 'No', 'Yes', 'Yes', 'No']
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Function to calculate Gini Impurity
def gini_impurity(labels):
    # Calculate the proportion of each class
    class_counts = labels.value_counts()
    total = len(labels)
    impurity = 1 - sum((count / total) ** 2 for count in class_counts)
    return impurity

# Function to calculate Gini Impurity for a feature split
def weighted_gini_impurity(data, feature, target):
    # Calculate the total Gini impurity before the split (root Gini Impurity)
    total_gini = gini_impurity(data[target])

    # Split the dataset by the feature
    feature_values = data[feature].unique()

    # Calculate the weighted Gini Impurity after the split
    weighted_impurity = 0
    for value in feature_values:
        subset = data[data[feature] == value]
        weighted_impurity += (len(subset) / len(data)) * gini_impurity(subset[target])

    return weighted_impurity

# 1. Gini Impurity for the whole dataset
total_gini = gini_impurity(df['Buys Product (Yes/No)'])
print(f"Gini Impurity for the entire dataset: {total_gini:.4f}")

# 2. Weighted Gini Impurity after splitting by Income Level
gini_after_split = weighted_gini_impurity(df, 'Income Level', 'Buys Product (Yes/No)')
print(f"Weighted Gini Impurity after splitting by Income Level: {gini_after_split:.4f}")
```
