```
import numpy as np
import matplotlib.pyplot as plt

# Example data for Gain Chart
deciles = np.arange(0, 110, 10)  # Percentile bins (0% to 100%)
gain_values = [0, 23.3, 43.3, 60, 73.3, 83.3, 90, 96.7, 100, 100, 100]  # Cumulative gain percentages

# Plot Gain Chart
plt.figure(figsize=(8, 5))
plt.plot(deciles, gain_values, marker=‘o’, linestyle=‘-’, color=‘b’, label="Model Gain Curve")
plt.plot(deciles, deciles, linestyle=‘--’, color=‘r’, label="Random Selection (Baseline)")

# Labels and Title
plt.xlabel("Percentage of Customers Targeted")
plt.ylabel("Cumulative Gain (%)")
plt.title("Gain Chart")
plt.legend()
plt.grid(True)
plt.show()
```
