```
# Prior probabilities
P_A = 0.6
P_B = 0.3
P_C = 0.1

# Conditional probabilities (Likelihoods)
P_S_given_A = 0.2
P_S_given_B = 0.7
P_S_given_C = 0.9

# Calculate P(S) using the law of total probability
P_S = (P_S_given_A * P_A) + (P_S_given_B * P_B) + (P_S_given_C * P_C)

# Calculate posterior probabilities using Bayes’ rule
P_A_given_S = (P_S_given_A * P_A) / P_S
P_B_given_S = (P_S_given_B * P_B) / P_S
P_C_given_S = (P_S_given_C * P_C) / P_S

# Print the results
print(f"P(S) (Probability of symptoms): {P_S}")
print(f"P(A|S) (Probability of Condition A given symptoms): {P_A_given_S:.3f}")  # Format to 3 decimal places
print(f"P(B|S) (Probability of Condition B given symptoms): {P_B_given_S:.3f}")
print(f"P(C|S) (Probability of Condition C given symptoms): {P_C_given_S:.3f}")

#Verification (posterior probabilities should sum to 1)
total_posterior_prob = P_A_given_S + P_B_given_S + P_C_given_S
print(f"\nSum of posterior probabilities: {total_posterior_prob:.3f}") #Should be very close to 1.0
```
