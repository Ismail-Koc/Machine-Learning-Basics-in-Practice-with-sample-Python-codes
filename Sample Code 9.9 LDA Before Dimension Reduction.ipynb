```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
import seaborn as sns
import pandas as pd

# Generate Synthetic Dataset with 3 Classes and 5 Features
X, y = make_classification(
    n_samples=300, n_features=5, n_informative=3, n_classes=3, random_state=42
)

# Convert to DataFrame for Pairwise Visualization
df = pd.DataFrame(X, columns=["Feature 1", "Feature 2", "Feature 3", "Feature 4", "Feature 5"])
df['Class'] = y  # Add class labels to the dataset

# Pairwise Scatter Plot of Features (Before LDA)
sns.pairplot(df, hue='Class', palette='viridis', corner=True)
plt.suptitle("Pairwise Scatter Plot of Features (Before Dimensionality Reduction)", fontsize=16)
plt.tight_layout()
plt.show()
```
