```
# Import necessary libraries
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.feature_selection import SequentialFeatureSelector

# Generate synthetic data for the example
# Create 100 samples with 4 features
X, y = make_regression(n_samples=100, n_features=4, noise=0.1, random_state=42)

# Initialize a linear regression model
model = LinearRegression()

# Forward selection: Select the top 2 features
sfs_forward = SequentialFeatureSelector(estimator=model, n_features_to_select=2, direction=‘forward’)
sfs_forward.fit(X, y)

# Get selected features (forward selection)
selected_features_forward = sfs_forward.get_support()
print("Selected Features (Forward Selection):", np.where(selected_features_forward)[0])

# Backward elimination: Select the top 2 features
sfs_backward = SequentialFeatureSelector(estimator=model, n_features_to_select=2, direction=‘backward’)
sfs_backward.fit(X, y)

# Get selected features (backward elimination)
selected_features_backward = sfs_backward.get_support()
print("Selected Features (Backward Elimination):", np.where(selected_features_backward)[0])
```
