```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# Step 1: Generate synthetic data
np.random.seed(42)
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=1.5, random_state=42)  # Unlabeled data

# Step 2: Initialize K-Means parameters
K = 3  # Number of clusters
max_iters = 5  # Number of iterations to demonstrate step-by-step
centroids = X[np.random.choice(X.shape[0], K, replace=False)]  # Randomly initialize K centroids

# Plot function to visualize each step
def plot_kmeans(X, centroids, labels=None, step=0):
    plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='coolwarm', alpha=0.6, edgecolors='k')
    plt.scatter(centroids[:, 0], centroids[:, 1], c='yellow', marker='X', s=200, label='Centroids')
    plt.title(f'K-Means Clustering - Step {step}')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.legend()
    plt.grid(alpha=0.5)
    plt.show()

# Step 3: Perform K-Means clustering manually (step by step)
for step in range(1, max_iters + 1):
    # Assign each point to the nearest centroid
    distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
    labels = np.argmin(distances, axis=1)

    # Plot current clustering state
    plot_kmeans(X, centroids, labels, step)

    # Compute new centroids (mean of assigned points)
    new_centroids = np.array([X[labels == k].mean(axis=0) for k in range(K)])

    # Check for convergence
    if np.all(centroids == new_centroids):
        print(f"Converged at Step {step}")
        break

    centroids = new_centroids  # Update centroids

print("Final cluster centroids:\n", centroids)
```
