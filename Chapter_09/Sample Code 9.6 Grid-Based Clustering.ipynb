```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from scipy.ndimage import measurements

# Step 1: Generate Sample Data
np.random.seed(42)
X, _ = make_blobs(n_samples=300, centers=[(2, 2), (6, 6), (10, 2)], cluster_std=1.0)

# Step 2: Define Grid Resolution
grid_size = 1.5  # Each grid cell has size 1.5x1.5
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

# Create grid
x_bins = np.arange(x_min, x_max, grid_size)
y_bins = np.arange(y_min, y_max, grid_size)

# Step 3: Assign Points to Grid Cells
grid_x = np.digitize(X[:, 0], bins=x_bins)
grid_y = np.digitize(X[:, 1], bins=y_bins)

# Create a 2D grid structure
grid = np.zeros((len(x_bins) + 1, len(y_bins) + 1))
for i in range(len(X)):
    grid[grid_x[i], grid_y[i]] += 1  # Count points in each grid cell

# Step 4: Cluster Grid Cells Using Density Threshold
density_threshold = 3  # Minimum number of points in a grid cell to form a cluster
grid_clusters, num_clusters = measurements.label(grid > density_threshold)

# Step 5: Assign Data Points to Clusters Based on Grid
point_labels = []
for i in range(len(X)):
    label = grid_clusters[grid_x[i], grid_y[i]]
    point_labels.append(label)

# Step 6: Plot the Grid and Clusters
plt.figure(figsize=(6, 6))
scatter = plt.scatter(X[:, 0], X[:, 1], c=point_labels, cmap="viridis", edgecolors='k', alpha=0.7)
plt.colorbar(scatter, label="Cluster ID")
plt.title("Grid-Based Clustering (STING)")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")

# Draw Grid
for x in x_bins:
    plt.axvline(x, color='gray', linestyle='dashed', alpha=0.5)
for y in y_bins:
    plt.axhline(y, color='gray', linestyle='dashed', alpha=0.5)

plt.grid(alpha=0.3)
plt.show()
```
