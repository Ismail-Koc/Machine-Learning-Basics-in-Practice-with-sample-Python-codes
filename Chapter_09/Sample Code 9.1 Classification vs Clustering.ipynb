```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score

# Generate synthetic dataset (3 clusters)
X, y = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.5)

# Split data for classification (supervised learning)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- Classification (Supervised Learning) ---
classifier = LogisticRegression()
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
classification_accuracy = accuracy_score(y_test, y_pred)

# --- Clustering (Unsupervised Learning) ---
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
y_cluster = kmeans.fit_predict(X)  # Unlabeled data, clustering based on similarity

# --- Visualization ---
plt.figure(figsize=(12, 5))

# Subplot 1: Classification Results
plt.subplot(1, 2, 1)
plt.scatter(X_test[:, 0], X_test[:, 1], c=y_pred, cmap='coolwarm', edgecolors='k')
plt.title(f'Classification (Supervised Learning)\nAccuracy: {classification_accuracy:.2f}')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(alpha=0.5)

# Subplot 2: Clustering Results
plt.subplot(1, 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=y_cluster, cmap='coolwarm', edgecolors='k')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            s=200, c='yellow', marker='X', label='Centroids')
plt.title('Clustering (Unsupervised Learning)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(alpha=0.5)

# Show plots
plt.tight_layout()
plt.show()
```
