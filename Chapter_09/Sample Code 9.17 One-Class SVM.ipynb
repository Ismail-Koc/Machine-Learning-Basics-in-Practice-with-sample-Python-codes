```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import OneClassSVM
from sklearn.datasets import make_blobs

# Step 1: Generate Synthetic Data
X, _ = make_blobs(n_samples=300, centers=1, cluster_std=1.5, random_state=42)
outliers = np.random.uniform(low=-10, high=10, size=(20, 2))
X = np.vstack([X, outliers])  # Combine normal data and outliers

# Step 2: Train One-Class SVM
model = OneClassSVM(kernel="rbf", gamma=0.1, nu=0.1)  # RBF kernel with parameters
model.fit(X)

# Step 3: Predict Anomalies
predictions = model.predict(X)  # -1: Anomaly, 1: Normal
normal_data = X[predictions == 1]
anomalous_data = X[predictions == -1]

# Step 4: Visualize Results
plt.figure(figsize=(10, 6))
plt.scatter(normal_data[:, 0], normal_data[:, 1], c="blue", label="Normal", alpha=0.7)
plt.scatter(anomalous_data[:, 0], anomalous_data[:, 1], c="red", label="Anomalies", alpha=0.7)
plt.title("One-Class SVM: Anomaly Detection", fontsize=16)
plt.xlabel("Feature 1", fontsize=12)
plt.ylabel("Feature 2", fontsize=12)
plt.legend()
plt.grid(alpha=0.4)
plt.show()
```
