```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs

# Step 1: Generate Sample Data (3 Gaussian Clusters)
np.random.seed(42)
X, _ = make_blobs(n_samples=300, centers=[(-2, -2), (3, 3), (7, -3)], cluster_std=[1.0, 1.2, 1.0])

# Step 2: Fit Gaussian Mixture Model (GMM) with 3 Components
gmm = GaussianMixture(n_components=3, covariance_type='full', random_state=42)
gmm.fit(X)
labels = gmm.predict(X)

# Step 3: Plot the Clustered Data Points
plt.figure(figsize=(6, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=30, edgecolors='k', alpha=0.7)
plt.title("Distribution-Based Clustering (GMM)")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")

# Step 4: Plot Gaussian Distributions (Ellipses)
import matplotlib.patches as patches

def draw_ellipse(position, covariance, ax, color):
    """Draw an ellipse representing a Gaussian component"""
    from matplotlib.transforms import Affine2D
    from scipy.linalg import eigh

    # Compute eigenvalues and eigenvectors
    U, s, Vt = np.linalg.svd(covariance)
    angle = np.degrees(np.arctan2(U[1, 0], U[0, 0]))
    width, height = 2 * np.sqrt(s)  # 2 standard deviations

    # Create an ellipse patch
    ellipse = patches.Ellipse(position, width, height, angle=angle, color=color, alpha=0.3)
    ax.add_patch(ellipse)

# Add ellipses for each Gaussian component
ax = plt.gca()
for i in range(gmm.n_components):
    draw_ellipse(gmm.means_[i], gmm.covariances_[i], ax, color='red')

plt.grid(alpha=0.5)
plt.show()
```
