```
import numpy as np
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
from scipy.spatial.distance import pdist

# Step 1: Generate 10 Random Data Points in 2D Space
np.random.seed(42)
X = np.random.rand(10, 2)  # 10 data points in 2D space

# Step 2: Compute the Distance Matrix and Apply Average Linkage
linkage_matrix = sch.linkage(X, method='average')  # "average" for Average Linkage

# Step 3: Plot the Dendrogram
plt.figure(figsize=(8, 5))
sch.dendrogram(linkage_matrix, labels=[f'P{i+1}' for i in range(len(X))], leaf_font_size=10)
plt.title("Hierarchical Clustering Dendrogram (Average Linkage)")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(alpha=0.5)
plt.show()

print(X)

plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c='blue', label='Random Points')

# Annotate each point with its index
for i, (x, y) in enumerate(X):
    plt.text(x, y, f'{i+1}', fontsize=16, ha='right')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Random Points Plot with Indices')

# Add a legend
plt.legend()

# Show the plot
plt.show()
```
